```python

"""
prune_qwen.py — скачивает Qwen2.5-0.5B, вырезает указанные слои,
сохраняет результат и прогоняет тест.

Запуск:
    pip install transformers torch accelerate
    python prune_qwen.py
"""

import os
import shutil
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer


# ----------------------- НАСТРОЙКИ -----------------------
MODEL_NAME = "Qwen/Qwen2.5-0.5B"
OUT_DIR = "./qwen05_pruned2"
N_KEEP = 12
# Режим удаления:
#   "tail"   — удалить N последних слоёв
#   "head"   — удалить N первых слоёв
#   "middle" — удалить список индексов
MODE = "middle"
N_REMOVE = 6              # для tail/head
DROP_INDICES = [10, 11, 12, 13, 14, 15]  # для middle

# Тестовые промпты
PROMPTS = [
    "Столица Франции —",
    "2 + 2 =",
    "Напиши короткое приветствие:",
]
MAX_NEW_TOKENS = 32
# ---------------------------------------------------------


def load_model():
    print(f"[1/5] Загружаю {MODEL_NAME} ...")
    tok = AutoTokenizer.from_pretrained(MODEL_NAME)
    model = AutoModelForCausalLM.from_pretrained(
        MODEL_NAME,
        torch_dtype=torch.float16,
        device_map="cpu",  # поменяй на "cuda" если есть GPU
    )
    model.eval()
    n = len(model.model.layers)
    print(f"      Слоёв в оригинале: {n}")
    return model, tok


def prune_layers(model, total, mode, n_remove, drop_indices=None, n_keep=None):
    print(f"[2/5] Вырезаю слои (режим={mode}) ...")
    layers = model.model.layers

    # --- какие индексы оставляем ---
    if mode == "tail":
        keep_indices = list(range(total - n_remove))
    elif mode == "head":
        keep_indices = list(range(n_remove, total))
    elif mode == "middle":
        drop = set(drop_indices or [])
        keep_indices = [i for i in range(total) if i not in drop]
    elif mode == "keep":
        keep_indices = list(range(n_keep))
    else:
        raise ValueError(f"Неизвестный режим: {mode}")

    # --- сами слои ---
    new_layers = [layers[i] for i in keep_indices]
    model.model.layers = torch.nn.ModuleList(new_layers)
    model.config.num_hidden_layers = len(new_layers)

    # --- ГЛАВНОЕ: переназначаем layer_idx у каждого слоя ---
    for new_idx, layer in enumerate(model.model.layers):
        if hasattr(layer, "self_attn") and hasattr(layer.self_attn, "layer_idx"):
            layer.self_attn.layer_idx = new_idx
        if hasattr(layer, "layer_idx"):
            layer.layer_idx = new_idx

    # --- синхронизируем layer_types ---
    if hasattr(model.config, "layer_types") and model.config.layer_types is not None:
        old_types = list(model.config.layer_types)
        if len(old_types) == total:
            model.config.layer_types = [old_types[i] for i in keep_indices]
            print(f"      layer_types обрезан до {len(model.config.layer_types)}")
        else:
            model.config.layer_types = None
            print(f"      layer_types сброшен")

    print(f"      Слоёв осталось: {len(new_layers)}")
    return model


def test_generate(model, tok, prompts, max_new_tokens=32):
    print("[3/5] Тестирую генерацию ...")
    device = next(model.parameters()).device
    for p in prompts:
        inputs = tok(p, return_tensors="pt").to(device)
        with torch.no_grad():
            out = model.generate(
                **inputs,
                max_new_tokens=max_new_tokens,
                do_sample=False,
                pad_token_id=tok.eos_token_id,
            )
        text = tok.decode(out[0], skip_special_tokens=True)
        print(f"      > {text!r}")


def save_model(model, tok, out_dir):
    print(f"[4/5] Сохраняю в {out_dir} ...")
    if os.path.exists(out_dir):
        shutil.rmtree(out_dir)
    os.makedirs(out_dir, exist_ok=True)
    model.save_pretrained(out_dir)
    tok.save_pretrained(out_dir)


def report_size(path):
    print("[5/5] Размер на диске:")
    total = 0
    for root, _, files in os.walk(path):
        for f in files:
            fp = os.path.join(root, f)
            size = os.path.getsize(fp)
            total += size
            print(f"      {size/1e6:8.2f} МБ  {os.path.relpath(fp, path)}")
    print(f"      ИТОГО: {total/1e6:.2f} МБ")


def main():
    model, tok = load_model()
    total = len(model.model.layers)
    model = prune_layers(
       model,
       total=total,
       mode=MODE,
       n_remove=N_REMOVE,
       drop_indices=DROP_INDICES,
       n_keep=N_KEEP,
    )
    test_generate(model, tok, PROMPTS, MAX_NEW_TOKENS)
    save_model(model, tok, OUT_DIR)
    report_size(OUT_DIR)
    print("\nГотово. Модель лежит в", OUT_DIR)


if __name__ == "__main__":
    main()
```

```python

"""
chat.py — интерактивный чат с локальной моделью Qwen.

Запуск:
    python chat.py

Команды:
    /exit  или  /quit   — выйти
    /clear              — очистить историю
    /reset              — сбросить всё к дефолту
    /temp 0.7           — изменить температуру
    /len 200            — изменить max_new_tokens
    /pen 1.3            — изменить repetition_penalty
"""

import os
os.environ["HF_ENDPOINT"] = "https://hf-mirror.com"

import torch
from transformers import AutoModelForCausalLM, AutoTokenizer


# ---------- НАСТРОЙКИ ----------
MODEL_PATH = "./qwen05_pruned2"
TOKENIZER_FALLBACK = "Qwen/Qwen2.5-0.5B"   # откуда брать токенизатор, если его нет в папке

MAX_HISTORY_TURNS = 6
MAX_NEW_TOKENS = 120
TEMPERATURE = 0.7
TOP_P = 0.9
REPETITION_PENALTY = 1.2

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
DTYPE = torch.float16 if DEVICE == "cuda" else torch.float32
# -------------------------------


def ensure_tokenizer(path):
    """
    Проверяет, есть ли в папке tokenizer.json.
    Если нет — докачивает токенизатор из оригинала и сохраняет в папку.
    """
    tok_json = os.path.join(path, "tokenizer.json")
    if os.path.exists(tok_json):
        return path

    print(f"[!] В {path} нет tokenizer.json — докачиваю из {TOKENIZER_FALLBACK}")
    try:
        tok = AutoTokenizer.from_pretrained(TOKENIZER_FALLBACK)
        tok.save_pretrained(path)
        print(f"[*] Токенизатор сохранён в {path}")
    except Exception as e:
        print(f"[!] Не удалось скачать токенизатор: {e}")
        print(f"[!] Буду грузить токенизатор напрямую из {TOKENIZER_FALLBACK}")
        return TOKENIZER_FALLBACK
    return path


def load_model(path):
    print(f"[*] Загружаю модель из {path} ...")
    if not os.path.isdir(path):
        raise FileNotFoundError(
            f"Папка {path} не найдена. Положи chat.py рядом с qwen05_pruned"
        )

    tok_path = ensure_tokenizer(path)
    tok = AutoTokenizer.from_pretrained(tok_path)

    model = AutoModelForCausalLM.from_pretrained(
        path,
        torch_dtype=DTYPE,
        device_map=DEVICE,
    )
    model.eval()

    n_layers = len(model.model.layers)
    print(f"[*] Модель загружена. Слоёв: {n_layers}. Устройство: {DEVICE}")
    print(f"[*] Команды: /exit  /clear  /reset  /temp X  /len N  /pen X")
    print("-" * 60)
    return model, tok


def build_prompt(history, user_msg):
    """Собирает промпт в формате Qwen-chat."""
    parts = []
    for role, text in history:
        if role == "user":
            parts.append(f"<|im_start|>user\n{text}<|im_end|>")
        else:
            parts.append(f"<|im_start|>assistant\n{text}<|im_end|>")
    parts.append(f"<|im_start|>user\n{user_msg}<|im_end|>")
    parts.append("<|im_start|>assistant\n")
    return "\n".join(parts)


def generate_reply(model, tok, prompt, max_new_tokens, temperature, top_p, rep_pen):
    inputs = tok(prompt, return_tensors="pt").to(DEVICE)

    do_sample = temperature > 0
    gen_kwargs = dict(
        max_new_tokens=max_new_tokens,
        do_sample=do_sample,
        repetition_penalty=rep_pen,
        pad_token_id=tok.eos_token_id,
    )
    if do_sample:
        gen_kwargs["temperature"] = temperature
        gen_kwargs["top_p"] = top_p

    with torch.no_grad():
        out = model.generate(**inputs, **gen_kwargs)

    new_tokens = out[0][inputs["input_ids"].shape[1]:]
    text = tok.decode(new_tokens, skip_special_tokens=True)

    for marker in ["<|im_end|>", "<|im_start|>", "<|endoftext|>"]:
        text = text.replace(marker, "")

    return text.strip()


def main():
    model, tok = load_model(MODEL_PATH)

    history = []
    max_new_tokens = MAX_NEW_TOKENS
    temperature = TEMPERATURE
    top_p = TOP_P
    rep_pen = REPETITION_PENALTY

    print("Готов. Пиши сообщение или команду.\n")

    while True:
        try:
            user = input("Ты: ").strip()
        except (EOFError, KeyboardInterrupt):
            print("\nВыход.")
            break

        if not user:
            continue

        # ----- команды -----
        if user in ("/exit", "/quit"):
            print("Выход.")
            break

        if user == "/clear":
            history = []
            print("[*] История очищена.\n")
            continue

        if user == "/reset":
            history = []
            max_new_tokens = MAX_NEW_TOKENS
            temperature = TEMPERATURE
            top_p = TOP_P
            rep_pen = REPETITION_PENALTY
            print("[*] Всё сброшено к дефолту.\n")
            continue

        if user.startswith("/temp "):
            try:
                temperature = float(user.split()[1])
                print(f"[*] temperature = {temperature}\n")
            except (IndexError, ValueError):
                print("[!] Пример: /temp 0.7\n")
            continue

        if user.startswith("/len "):
            try:
                max_new_tokens = int(user.split()[1])
                print(f"[*] max_new_tokens = {max_new_tokens}\n")
            except (IndexError, ValueError):
                print("[!] Пример: /len 200\n")
            continue

        if user.startswith("/pen "):
            try:
                rep_pen = float(user.split()[1])
                print(f"[*] repetition_penalty = {rep_pen}\n")
            except (IndexError, ValueError):
                print("[!] Пример: /pen 1.3\n")
            continue

        # ----- обычное сообщение -----
        prompt = build_prompt(history, user)

        try:
            reply = generate_reply(
                model, tok, prompt,
                max_new_tokens=max_new_tokens,
                temperature=temperature,
                top_p=top_p,
                rep_pen=rep_pen,
            )
        except Exception as e:
            print(f"[!] Ошибка генерации: {e}\n")
            continue

        if not reply:
            reply = "(пустой ответ — попробуй /clear или /temp 0.9)"

        print(f"Бот: {reply}\n")

        history.append(("user", user))
        history.append(("assistant", reply))
        if len(history) > MAX_HISTORY_TURNS * 2:
            history = history[-MAX_HISTORY_TURNS * 2:]


if __name__ == "__main__":
    main()
```
